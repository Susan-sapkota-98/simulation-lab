```python
import random
import matplotlib.pyplot as plt
x = 0
y = 0
x_points = [x]
y_points = [y]
for i in range(100):
    move = random.choices(
        ["forward", "left", "right"],
        weights=[50, 20, 30]
    )[0]
    if move == "forward":
        y += 1
    elif move == "left":
        x -= 1
    elif move == "right":
        x += 1
    x_points.append(x)
    y_points.append(y)
plt.figure(figsize=(7,7))
plt.plot(x_points, y_points, marker="o")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Random Person Walk")
plt.grid()
plt.show()
```
