```python
import numpy as np
import matplotlib.pyplot as plt
m = 1
k = 10
c = 0
x0 = 1
v0 = 0
t0, tf = 0, 20
h = 0.01 
n_steps = int((tf - t0) / h)

# Arrays to store results
t = np.linspace(t0, tf, n_steps + 1)
x = np.zeros(n_steps + 1)
v = np.zeros(n_steps + 1)
x[0], v[0] = x0, v0

# Derivative function: returns [dx/dt, dv/dt]
def f(x, v):
    dxdt = v
    dvdt = (-k * x - c * v) / m
    return dxdt, dvdt

# RK2 (Heun's method) integration loop
for i in range(n_steps):
    k1x, k1v = f(x[i], v[i])
    k2x, k2v = f(x[i] + h * k1x, v[i] + h * k1v)

    x[i+1] = x[i] + (h / 2) * (k1x + k2x)
    v[i+1] = v[i] + (h / 2) * (k1v + k2v)

# Plot results
plt.figure(figsize=(8, 4))
plt.plot(t, x, label="Displacement x(t)")
plt.xlabel("Time (s)")
plt.ylabel("Displacement (m)")
plt.title("Spring-Mass System (RK2 Method)")
plt.legend()
plt.grid(True)
plt.show()
```
