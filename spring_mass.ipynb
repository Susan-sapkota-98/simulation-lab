```python
import numpy as np
import matplotlib.pyplot as plt

# Ellipse parameters
semi_major = 5
semi_minor = 3
num_points = 10000

# Generate random points in bounding rectangle
px = np.random.uniform(-semi_major, semi_major, num_points)
py = np.random.uniform(-semi_minor, semi_minor, num_points)

# Check which points fall inside the ellipse
inside_mask = (px**2 / semi_major**2) + (py**2 / semi_minor**2) <= 1
hits = np.sum(inside_mask)

# Estimate area of ellipse
area_estimate = (4 * semi_major * semi_minor) * hits / num_points
print(f"Estimated ellipse area: {area_estimate:.4f}")

# Estimate pi from the same points
pi_estimate = 4 * hits / num_points
print(f"Estimated pi: {pi_estimate:.4f}")
print(f"True pi: {np.pi:.4f}")
print(f"Absolute error: {abs(pi_estimate - np.pi):.4f}")

# Plot
plt.figure(figsize=(6, 6))
plt.scatter(px[inside_mask], py[inside_mask], s=2, color="teal", label="Inside")
plt.scatter(px[~inside_mask], py[~inside_mask], s=2, color="lightgray", label="Outside")
plt.axis("equal")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Monte Carlo Estimation: Ellipse Area & Pi")
plt.legend()
plt.show()
```
