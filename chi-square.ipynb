```python
import random
faces = [1, 2, 3, 4, 5, 6]
probabilities = [1/6] * 6
n = 600
observed = [0] * 6
for _ in range(n):
    roll = random.randint(0, 5)
    observed[roll] += 1
expected = [n * p for p in probabilities]
chi_square = sum((observed[i] - expected[i]) ** 2 / expected[i] for i in range(6))
df = len(faces) - 1
critical_value = 11.070
print("Face:     ", faces)
print("Observed: ", observed)
print("Expected: ", [round(e, 2) for e in expected])
print(f"\nChi-Square Statistic = {chi_square:.4f}")
print(f"Degrees of Freedom   = {df}")
print(f"Critical Value (α=0.05) = {critical_value}")

if chi_square > critical_value:
    print("\nResult: Reject H0 -> The die appears to be biased.")
else:
    print("\nResult: Fail to reject H0 -> The die appears to be fair.")
```
