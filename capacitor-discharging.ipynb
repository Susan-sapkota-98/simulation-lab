```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
R = 1000          # Resistance (Ohms)
C = 0.001         # Capacitance (Farads)
V0 = 10           # Initial capacitor voltage (Volts)

# Simulation settings
h = 0.1           # Time step
t_max = 5         # Total simulation time

# Differential equation
def f(t, V):
    return -V / (R * C)

# Initialization
t = 0
V = V0

time = [t]
voltage = [V]

# RK2 Method
while t < t_max:
    k1 = h * f(t, V)
    k2 = h * f(t + h/2, V + k1/2)

    V = V + k2
    t = t + h

    time.append(t)
    voltage.append(V)

# Print results
print("Time\tVoltage")
for i in range(len(time)):
    print(f"{time[i]:.2f}\t{voltage[i]:.4f}")

# Plot
plt.plot(time, voltage, marker='o')
plt.xlabel("Time (s)")
plt.ylabel("Capacitor Voltage (V)")
plt.title("Capacitor Discharge using RK2 Method")
plt.grid(True)
plt.show()
```
